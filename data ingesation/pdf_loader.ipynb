```python
import asyncio
import nest_asyncio
from langchain_community.document_loaders import PyPDFLoader
nest_asyncio.apply()
async def main():
    loader = PyPDFLoader("resume.pdf")
    docs= await loader.aload()
    for doc in docs:
        print(doc)

asyncio.run(main())
```
